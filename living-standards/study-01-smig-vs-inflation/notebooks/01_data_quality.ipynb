```python
import pandas as pd
import numpy as np

from pathlib import Path

import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)

plt.style.use("default")
```

## Project Configuration

```python
# -----------------------------------------------------------------------------
# Project Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path.cwd().parent

RAW_DATA = PROJECT_ROOT / "data" / "raw"
PROCESSED_DATA = PROJECT_ROOT / "data" / "processed"
OUTPUTS = PROJECT_ROOT / "outputs"

RAW_FILE = RAW_DATA / "smig_cpi_2000_2024.xlsx"

print(PROJECT_ROOT)
print(RAW_FILE)
```

## Load Dataset

```python
df = pd.read_excel(RAW_FILE)

df.head()
```

## Dataset Overview

```python
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

df.info()
```

## Data Types

```python
df.dtypes
```

## Missing Values

```python
missing = df.isna().sum()

missing
```

```python
missing_percentage = df.isna().mean() * 100

missing_percentage
```

## Duplicate Rows

```python
duplicates = df.duplicated().sum()

print(f"Duplicate rows: {duplicates}")
```

## Year Validation

```python
years = df["year"]

print(years.min())
print(years.max())

print(years.is_monotonic_increasing)
```

```python
expected_years = list(range(2000, 2025))

missing_years = sorted(
    set(expected_years) - set(df["year"])
)

missing_years
```

## Descriptive Statistics

```python
df.describe().T
```

## Export Clean Dataset

```python
output_file = PROCESSED_DATA / "smig_cpi_2000_2024_clean.csv"

df.to_csv(output_file, index=False)

print("Dataset exported successfully.")
```

## Data Quality Report

```python
print("=" * 60)
print("DATA QUALITY REPORT")
print("=" * 60)

print()

print(f"Observations      : {len(df)}")
print(f"Variables         : {len(df.columns)}")
print(f"Missing values    : {df.isna().sum().sum()}")
print(f"Duplicate rows    : {df.duplicated().sum()}")
print(f"Start year        : {df['year'].min()}")
print(f"End year          : {df['year'].max()}")

print()

print("Dataset Status")

if (
    df.isna().sum().sum() == 0
    and df.duplicated().sum() == 0
):
    print("PASS - Dataset ready for feature engineering.")
else:
    print("WARNING - Dataset requires additional cleaning.")
```
