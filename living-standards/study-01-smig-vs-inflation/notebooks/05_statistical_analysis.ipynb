## Has Morocco's Minimum Wage Protected Workers Against Inflation Since 2000?

---

## Objective

This notebook produces publication-quality figures for reports, presentations and LinkedIn publications.

Each visualization is exported in high resolution.

```python
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path

plt.rcParams["figure.figsize"] = (12,7)

plt.rcParams["figure.dpi"] = 150

plt.rcParams["font.size"] = 12

plt.rcParams["axes.titleweight"] = "bold"

plt.rcParams["axes.grid"] = True
```

```python
PROJECT_ROOT = Path.cwd().parent

DATA = PROJECT_ROOT / "data" / "processed"

OUTPUT = PROJECT_ROOT / "outputs"

FIGURES = OUTPUT / "figures"

FIGURES.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(DATA / "smig_cpi_features.csv")
```

## Figure 1

Monthly SMIG vs CPI

```python
plt.figure()

plt.plot(
    df["year"],
    df["smig_monthly"],
    linewidth=3,
    label="Monthly SMIG"
)

plt.plot(
    df["year"],
    df["cpi"],
    linewidth=3,
    label="Consumer Price Index"
)

plt.title("Monthly Minimum Wage and Consumer Price Index")

plt.xlabel("Year")

plt.legend()

plt.tight_layout()

plt.savefig(
    FIGURES/"figure_01_smig_vs_cpi.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## Figure 2

Base 100 Comparison

```python
plt.figure()

plt.plot(
    df["year"],
    df["SMIG_Index"],
    linewidth=3,
    label="SMIG Index"
)

plt.plot(
    df["year"],
    df["CPI_Index"],
    linewidth=3,
    label="CPI Index"
)

plt.title("Indexed Evolution (Base 100 in 2000)")

plt.xlabel("Year")

plt.ylabel("Index")

plt.legend()

plt.tight_layout()

plt.savefig(
    FIGURES/"figure_02_base100.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## Figure 3

Annual Inflation

```python
plt.figure()

plt.bar(
    df["year"],
    df["Inflation_Rate"]
)

plt.title("Annual Inflation Rate")

plt.xlabel("Year")

plt.ylabel("%")

plt.tight_layout()

plt.savefig(
    FIGURES/"figure_03_inflation.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## Figure 4

Nominal Wage Growth

```python
plt.figure()

plt.bar(
    df["year"],
    df["Nominal_Wage_Growth"]
)

plt.title("Annual Nominal Wage Growth")

plt.xlabel("Year")

plt.ylabel("%")

plt.tight_layout()

plt.savefig(
    FIGURES/"figure_04_wage_growth.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## Figure 5

Real Wage Index

```python
plt.figure()

plt.plot(
    df["year"],
    df["Real_Wage_Index"],
    linewidth=3
)

plt.title("Real Wage Index")

plt.xlabel("Year")

plt.ylabel("Index")

plt.tight_layout()

plt.savefig(
    FIGURES/"figure_05_real_wage.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## Figure 6

Purchasing Power Gap

```python
plt.figure()

plt.bar(
    df["year"],
    df["Purchasing_Power_Gap"]
)

plt.axhline(
    y=0,
    linestyle="--",
    linewidth=2
)

plt.title("Purchasing Power Gap")

plt.xlabel("Year")

plt.ylabel("Percentage Points")

plt.tight_layout()

plt.savefig(
    FIGURES/"figure_06_gap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## Figure 7

Purchasing Power Protection Ratio

```python
plt.figure()

plt.plot(
    df["year"],
    df["PPPR"],
    linewidth=3
)

plt.axhline(
    y=1,
    linestyle="--",
    linewidth=2
)

plt.title("Purchasing Power Protection Ratio")

plt.xlabel("Year")

plt.ylabel("Ratio")

plt.tight_layout()

plt.savefig(
    FIGURES/"figure_07_pppr.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## Export Summary

```python
print("="*70)

print("PUBLICATION FIGURES GENERATED")

print("="*70)

for file in sorted(FIGURES.glob("*.png")):

    print(file.name)

print()

print("All publication figures successfully exported.")
```
