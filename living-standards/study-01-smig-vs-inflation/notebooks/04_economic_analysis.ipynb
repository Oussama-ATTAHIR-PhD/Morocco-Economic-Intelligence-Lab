## Objective

This notebook transforms engineered indicators into economic insights.

It produces quantitative evidence that answers the research question through descriptive economic analytics.

```python
import pandas as pd
import numpy as np

from pathlib import Path

pd.set_option("display.max_columns", None)
```

## Load Feature Dataset

```python
PROJECT_ROOT = Path.cwd().parent

DATA = PROJECT_ROOT / "data" / "processed"

df = pd.read_csv(DATA / "smig_cpi_features.csv")

df.head()
```

## Study Period

```python
START_YEAR = df["year"].min()
END_YEAR = df["year"].max()

print(f"Study Period : {START_YEAR}-{END_YEAR}")
```

## Economic Indicators

```python
average_inflation = df["Inflation_Rate"].mean()

average_nominal_growth = df["Nominal_Wage_Growth"].mean()

average_real_growth = df["Real_Wage_Growth"].mean()

average_pppr = df["PPPR"].mean()

print("Average Inflation :", round(average_inflation,2))

print("Average Wage Growth :", round(average_nominal_growth,2))

print("Average Real Growth :", round(average_real_growth,2))

print("Average PPPR :", round(average_pppr,3))
```

## Purchasing Power Performance

```python
years_above = (
    df["Purchasing_Power_Gap"] > 0
).sum()

years_below = (
    df["Purchasing_Power_Gap"] < 0
).sum()

years_equal = (
    df["Purchasing_Power_Gap"] == 0
).sum()

print(years_above)

print(years_below)

print(years_equal)
```

## Best and Worst Years

```python
best_year = df.loc[
    df["Purchasing_Power_Gap"].idxmax()
]

worst_year = df.loc[
    df["Purchasing_Power_Gap"].idxmin()
]

best_year
```

```python
worst_year
```

## Cumulative Performance

```python
cumulative_nominal = df["Cumulative_Wage_Growth"].iloc[-1]

cumulative_inflation = df["Cumulative_Inflation"].iloc[-1]

cumulative_real = (
    cumulative_nominal -
    cumulative_inflation
)

print(round(cumulative_nominal,2))

print(round(cumulative_inflation,2))

print(round(cumulative_real,2))
```

## Volatility

```python
inflation_volatility = df["Inflation_Rate"].std()

wage_volatility = df["Nominal_Wage_Growth"].std()

print(round(inflation_volatility,2))

print(round(wage_volatility,2))
```

## Decade Analysis

```python
df["decade"] = (df["year"] // 10) * 10

decade_summary = (
    df
    .groupby("decade")
    [["Inflation_Rate",
      "Nominal_Wage_Growth",
      "Real_Wage_Growth"]]
    .mean()
    .round(2)
)

decade_summary
```

## Economic Diagnostics

```python
real_positive = (
    df["Real_Wage_Growth"] > 0
).sum()

real_negative = (
    df["Real_Wage_Growth"] < 0
).sum()

print(real_positive)

print(real_negative)
```

## Executive Indicators

```python
study_results = {

    "study_period":f"{START_YEAR}-{END_YEAR}",

    "observations":len(df),

    "average_inflation":round(average_inflation,2),

    "average_nominal_growth":round(average_nominal_growth,2),

    "average_real_growth":round(average_real_growth,2),

    "average_pppr":round(average_pppr,3),

    "years_above_inflation":int(years_above),

    "years_below_inflation":int(years_below),

    "years_equal":int(years_equal),

    "best_year":int(best_year["year"]),

    "worst_year":int(worst_year["year"]),

    "best_gap":round(best_year["Purchasing_Power_Gap"],2),

    "worst_gap":round(worst_year["Purchasing_Power_Gap"],2),

    "cumulative_real_gain":round(cumulative_real,2)

}

study_results
```

## Automatically Generated Key Findings

```python
key_findings = []

if years_above > years_below:

    key_findings.append(
        "Minimum wage outpaced inflation during most years."
    )

else:

    key_findings.append(
        "Inflation outpaced minimum wage during most years."
    )

if cumulative_real > 0:

    key_findings.append(
        "The cumulative purchasing power improved over the study period."
    )

else:

    key_findings.append(
        "The cumulative purchasing power deteriorated over the study period."
    )

if average_pppr > 1:

    key_findings.append(
        "On average, minimum wage growth exceeded inflation."
    )

else:

    key_findings.append(
        "On average, inflation exceeded minimum wage growth."
    )

for finding in key_findings:

    print("•", finding)
```

## Export Study Results

```python
results = pd.DataFrame([study_results])

results.to_csv(

    DATA / "study_results.csv",

    index=False

)

results
```

## Final Assessment

```python
print("="*70)

print("ECONOMIC ANALYSIS COMPLETED")

print("="*70)

print()

print(f"Study Period : {study_results['study_period']}")

print(f"Average Inflation : {study_results['average_inflation']} %")

print(f"Average Wage Growth : {study_results['average_nominal_growth']} %")

print(f"Average Real Growth : {study_results['average_real_growth']} %")

print(f"Years Above Inflation : {study_results['years_above_inflation']}")

print(f"Years Below Inflation : {study_results['years_below_inflation']}")

print()

print("Dataset ready for publication-quality visualization.")
```
