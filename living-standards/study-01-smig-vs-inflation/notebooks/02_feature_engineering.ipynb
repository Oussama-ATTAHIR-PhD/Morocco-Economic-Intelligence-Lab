- Real Wage Index
- Real Wage Growth

- Purchasing Power Protection Ratio (PPPR)

The resulting dataset will be exported for subsequent analyses.

```python
import pandas as pd
import numpy as np

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
```

## Project Configuration

```python
# -----------------------------------------------------------------------------
# Project Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path.cwd().parent

PROCESSED_DATA = PROJECT_ROOT / "data" / "processed"

INPUT_FILE = PROCESSED_DATA / "smig_cpi_2000_2024_clean.csv"

OUTPUT_FILE = PROCESSED_DATA / "smig_cpi_features.csv"

print(INPUT_FILE)
print(OUTPUT_FILE)
```

## Load Dataset

```python
df = pd.read_csv(INPUT_FILE)

df.head()
```

## Base Indices (2000 = 100)

```python
base_smig = df.loc[df["year"] == 2000, "smig_monthly"].iloc[0]

base_cpi = df.loc[df["year"] == 2000, "cpi"].iloc[0]

print(base_smig)

print(base_cpi)
```

```python
df["SMIG_Index"] = (
    df["smig_monthly"] / base_smig
) * 100

df["CPI_Index"] = (
    df["cpi"] / base_cpi
) * 100

df.head()
```

## Annual Inflation Rate

```python
df["Inflation_Rate"] = (
    df["cpi"].pct_change()
) * 100
```

## Nominal Wage Growth

```python
df["Nominal_Wage_Growth"] = (
    df["smig_monthly"].pct_change()
) * 100
```

## Real Wage Index

```python
df["Real_Wage_Index"] = (
    df["SMIG_Index"] /
    df["CPI_Index"]
) * 100
```

## Real Wage Growth

```python
df["Real_Wage_Growth"] = (
    df["Nominal_Wage_Growth"] -
    df["Inflation_Rate"]
)
```

## Purchasing Power Gap

```python
df["Purchasing_Power_Gap"] = (
    df["Nominal_Wage_Growth"] -
    df["Inflation_Rate"]
)
```

## Purchasing Power Protection Ratio (PPPR)

```python
df["PPPR"] = (
    df["SMIG_Index"] /
    df["CPI_Index"]
)
```

## Cumulative Inflation

```python
df["Cumulative_Inflation"] = (
    df["cpi"] /
    base_cpi -
    1
) * 100
```

## Cumulative Wage Growth

```python
df["Cumulative_Wage_Growth"] = (
    df["smig_monthly"] /
    base_smig -
    1
) * 100
```

## Preview Feature Dataset

```python
df.round(2)
```

## Feature Validation

```python
print("=" * 70)

print("FEATURE ENGINEERING REPORT")

print("=" * 70)

print()

print(f"Rows                 : {len(df)}")

print(f"Columns              : {len(df.columns)}")

print()

print("Generated Features")

generated = [

    "SMIG_Index",

    "CPI_Index",

    "Inflation_Rate",

    "Nominal_Wage_Growth",

    "Real_Wage_Index",

    "Real_Wage_Growth",

    "Purchasing_Power_Gap",

    "PPPR",

    "Cumulative_Inflation",

    "Cumulative_Wage_Growth"

]

for feature in generated:

    print(f"✓ {feature}")

print()

print("Feature engineering completed successfully.")
```

## Export Dataset

```python
df.to_csv(

    OUTPUT_FILE,

    index=False

)

print("Feature dataset exported successfully.")
```
