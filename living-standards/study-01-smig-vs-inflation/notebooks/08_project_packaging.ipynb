# Study 01 — Project Packaging

## Has Morocco's Minimum Wage Protected Workers Against Inflation Since 2000?

### Morocco Economic Intelligence Lab (MEIL)

---

## Objective

This notebook packages the study outputs and prepares the project for GitHub publication.

It validates the repository structure, checks the generated artefacts, and creates a final study manifest.

```python
import pandas as pd
from pathlib import Path
from datetime import datetime
import subprocess
from IPython.display import display
```

```python
NOTEBOOKS_DIR = Path.cwd()
STUDY_ROOT = NOTEBOOKS_DIR.parent
REPO_ROOT = STUDY_ROOT.parents[1]

DATA_DIR = STUDY_ROOT / "data"
RAW_DIR = DATA_DIR / "raw"
PROCESSED_DIR = DATA_DIR / "processed"

OUTPUTS_DIR = STUDY_ROOT / "outputs"
FIGURES_DIR = OUTPUTS_DIR / "figures"
TABLES_DIR = OUTPUTS_DIR / "tables"

EXEC_SUMMARY_DIR = STUDY_ROOT / "executive-summary"

FEATURE_FILE = PROCESSED_DIR / "smig_cpi_features.csv"
RESULTS_FILE = PROCESSED_DIR / "study_results.csv"
CLEAN_FILE = PROCESSED_DIR / "smig_cpi_2000_2024_clean.csv"
REPORT_FILE = EXEC_SUMMARY_DIR / "study_01_executive_summary.md"
PACKAGE_MANIFEST = STUDY_ROOT / "study_01_package_manifest.md"

print("Notebook directory :", NOTEBOOKS_DIR)
print("Study root         :", STUDY_ROOT)
print("Repo root          :", REPO_ROOT)
```

## Expected Repository Structure

```python
expected_repo_paths = [
    REPO_ROOT / "README.md",
    REPO_ROOT / "LICENSE",
    REPO_ROOT / "requirements.txt",
    REPO_ROOT / ".gitignore",
    REPO_ROOT / "living-standards",
]

expected_study_paths = [
    STUDY_ROOT / "README.md",
    STUDY_ROOT / "methodology",
    STUDY_ROOT / "references",
    STUDY_ROOT / "data" / "raw",
    STUDY_ROOT / "data" / "processed",
    STUDY_ROOT / "notebooks",
    STUDY_ROOT / "outputs" / "figures",
    STUDY_ROOT / "outputs" / "tables",
    STUDY_ROOT / "executive-summary",
]

missing_repo_paths = [str(path) for path in expected_repo_paths if not path.exists()]
missing_study_paths = [str(path) for path in expected_study_paths if not path.exists()]

print("Repo-level checks")
if missing_repo_paths:
    for path in missing_repo_paths:
        print(" -", path)
else:
    print("All repo-level paths are present.")

print()
print("Study-level checks")
if missing_study_paths:
    for path in missing_study_paths:
        print(" -", path)
else:
    print("All study-level paths are present.")
```

## Study Artefacts Overview

```python
artefacts = pd.DataFrame([
    {"File": "smig_cpi_2000_2024_clean.csv", "Exists": CLEAN_FILE.exists(), "Path": str(CLEAN_FILE)},
    {"File": "smig_cpi_features.csv", "Exists": FEATURE_FILE.exists(), "Path": str(FEATURE_FILE)},
    {"File": "study_results.csv", "Exists": RESULTS_FILE.exists(), "Path": str(RESULTS_FILE)},
    {"File": "study_01_executive_summary.md", "Exists": REPORT_FILE.exists(), "Path": str(REPORT_FILE)},
])

display(artefacts)
```

## Figures Inventory

```python
if FIGURES_DIR.exists():
    figure_files = sorted(FIGURES_DIR.glob("*.png"))
    figures_df = pd.DataFrame([
        {
            "Figure": file.name,
            "Exists": file.exists(),
            "Size_KB": round(file.stat().st_size / 1024, 2)
        }
        for file in figure_files
    ])
else:
    figures_df = pd.DataFrame(columns=["Figure", "Exists", "Size_KB"])

display(figures_df)
```

## Tables Inventory

```python
if TABLES_DIR.exists():
    table_files = sorted(TABLES_DIR.glob("*"))
    if table_files:
        tables_df = pd.DataFrame([
            {
                "File": file.name,
                "Exists": file.exists(),
                "Size_KB": round(file.stat().st_size / 1024, 2) if file.is_file() else None
            }
            for file in table_files
        ])
        display(tables_df)
    else:
        print("No table files found in outputs/tables.")
else:
    print("The tables directory does not exist.")
```

## Git Status

```python
try:
    result = subprocess.run(
        ["git", "status", "--short"],
        cwd=REPO_ROOT,
        capture_output=True,
        text=True,
        check=False
    )

    if result.stdout.strip():
        print(result.stdout)
    else:
        print("Working tree clean or no tracked changes yet.")
except FileNotFoundError:
    print("Git is not available in this environment.")
```

## Study Statistics

```python
results = pd.read_csv(RESULTS_FILE).iloc[0].to_dict()

summary_stats = pd.DataFrame([
    {"Metric": "Study period", "Value": results.get("study_period")},
    {"Metric": "Observations", "Value": results.get("observations")},
    {"Metric": "Average inflation (%)", "Value": results.get("average_inflation")},
    {"Metric": "Average nominal wage growth (%)", "Value": results.get("average_nominal_growth")},
    {"Metric": "Average real wage growth (%)", "Value": results.get("average_real_growth")},
    {"Metric": "Average PPPR", "Value": results.get("average_pppr")},
    {"Metric": "Years above inflation", "Value": results.get("years_above_inflation")},
    {"Metric": "Years below inflation", "Value": results.get("years_below_inflation")},
    {"Metric": "Best year", "Value": results.get("best_year")},
    {"Metric": "Worst year", "Value": results.get("worst_year")},
    {"Metric": "Cumulative real gain (%)", "Value": results.get("cumulative_real_gain")},
])

display(summary_stats)
```

## Build Study Manifest

```python
def bullet_list(items):
    return "\n".join([f"- {item}" for item in items])

manifest_lines = []

manifest_lines.append("# Study 01 — Package Manifest")
manifest_lines.append("")
manifest_lines.append("## Has Morocco's Minimum Wage Protected Workers Against Inflation Since 2000?")
manifest_lines.append("")
manifest_lines.append("### Morocco Economic Intelligence Lab (MEIL)")
manifest_lines.append("")
manifest_lines.append(f"Generated on: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
manifest_lines.append("")
manifest_lines.append("## Core Artefacts")
manifest_lines.append("")
manifest_lines.append(bullet_list([
    f"smig_cpi_2000_2024_clean.csv — {'present' if CLEAN_FILE.exists() else 'missing'}",
    f"smig_cpi_features.csv — {'present' if FEATURE_FILE.exists() else 'missing'}",
    f"study_results.csv — {'present' if RESULTS_FILE.exists() else 'missing'}",
    f"study_01_executive_summary.md — {'present' if REPORT_FILE.exists() else 'missing'}",
]))
manifest_lines.append("")
manifest_lines.append("## Figures")
manifest_lines.append("")

if not figures_df.empty:
    for _, row in figures_df.iterrows():
        manifest_lines.append(f"- {row['Figure']} ({row['Size_KB']} KB)")
else:
    manifest_lines.append("- No figures found.")

manifest_lines.append("")
manifest_lines.append("## Key Study Statistics")
manifest_lines.append("")
for _, row in summary_stats.iterrows():
    manifest_lines.append(f"- {row['Metric']}: {row['Value']}")

manifest_lines.append("")
manifest_lines.append("## Repository Notes")
manifest_lines.append("")
manifest_lines.append("- The study follows a modular notebook workflow.")
manifest_lines.append("- All outputs are stored inside the study folder.")
manifest_lines.append("- The report is ready for GitHub publication.")
manifest_lines.append("")
manifest_lines.append("## Publication Checklist")
manifest_lines.append("")
manifest_lines.append("- [x] Data quality validated")
manifest_lines.append("- [x] Feature engineering completed")
manifest_lines.append("- [x] Exploratory analysis completed")
manifest_lines.append("- [x] Economic analysis completed")
manifest_lines.append("- [x] Statistical analysis completed")
manifest_lines.append("- [x] Publication-quality figures generated")
manifest_lines.append("- [x] Executive summary exported")
manifest_lines.append("- [x] Package manifest created")

package_md = "\n".join(manifest_lines)

print(package_md[:4000])
```

```python
with open(PACKAGE_MANIFEST, "w", encoding="utf-8") as f:
    f.write(package_md)

print(f"Package manifest exported successfully to: {PACKAGE_MANIFEST}")
```

## Final Publication Readiness Check

```python
required_files = [
    CLEAN_FILE,
    FEATURE_FILE,
    RESULTS_FILE,
    REPORT_FILE,
    PACKAGE_MANIFEST,
]

required_figures = [
    FIGURES_DIR / "figure_01_executive_dashboard.png",
    FIGURES_DIR / "figure_02_base100_comparison.png",
    FIGURES_DIR / "figure_03_purchasing_power_gap.png",
    FIGURES_DIR / "figure_04_pppr.png",
    FIGURES_DIR / "figure_05_inflation_vs_wage_growth.png",
    FIGURES_DIR / "figure_06_cumulative_growth.png",
    FIGURES_DIR / "figure_07_revaluation_timeline.png",
    FIGURES_DIR / "figure_08_economic_heatmap.png",
]

missing_required = [str(path) for path in required_files if not path.exists()]
missing_figures = [str(path) for path in required_figures if not path.exists()]

print("=" * 70)
print("PUBLICATION READINESS CHECK")
print("=" * 70)
print()

if missing_required:
    print("Missing required files:")
    for item in missing_required:
        print(" -", item)
else:
    print("All required study files are present.")

print()

if missing_figures:
    print("Missing required figures:")
    for item in missing_figures:
        print(" -", item)
else:
    print("All required publication figures are present.")

print()

if not missing_required and not missing_figures:
    print("STATUS: READY FOR GITHUB PUBLICATION")
else:
    print("STATUS: NOT READY — fix missing artefacts first.")
```

## Package Summary

```python
print("Project packaging completed successfully.")
print()
print(f"Manifest file : {PACKAGE_MANIFEST}")
print(f"Figures folder : {FIGURES_DIR}")
print(f"Report folder  : {EXEC_SUMMARY_DIR}")
print()
print("Study 01 is ready for final GitHub cleanup and publication.")
```
