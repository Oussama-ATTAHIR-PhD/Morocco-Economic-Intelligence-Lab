```python
# ============================================================
# 02 - FEATURE ENGINEERING
# HOUSEHOLD PURCHASING POWER IN MOROCCO
# PERIOD: 2007-2024
# ============================================================

import pandas as pd
import numpy as np

pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

file_path = (
    "../data/processed/"
    "study_02_household_purchasing_power_2007_2024.xlsx"
)

df = pd.read_excel(
    file_path,
    sheet_name="master_data"
)

print("File loaded successfully.")
print("Shape:", df.shape)

display(df.head())
```

```python
# ============================================================
# VERIFY AND RESTRICT STUDY PERIOD
# ============================================================

# Convert year to numeric
df["year"] = pd.to_numeric(df["year"], errors="coerce")

# Keep only the official study period
df = df[
    (df["year"] >= 2007) &
    (df["year"] <= 2024)
].copy()

# Convert year to integer
df["year"] = df["year"].astype(int)

# Sort chronologically
df = df.sort_values("year").reset_index(drop=True)

# Validation
expected_columns = [
    "year",
    "cpi_base_2017_harmonized",
    "disposable_income_s14_harmonized",
    "consumption_s14_harmonized"
]

assert df.columns.tolist() == expected_columns, (
    "Unexpected column structure"
)

assert df["year"].min() == 2007
assert df["year"].max() == 2024
assert len(df) == 18

assert df.isna().sum().sum() == 0, (
    "Missing values detected"
)

print("Dataset structure: VALID")
print("Shape:", df.shape)
print("Period:", df["year"].min(), "-", df["year"].max())
print("Missing values:", df.isna().sum().sum())

display(df)
```

```python
# ============================================================
# CREATE ANALYTICAL VARIABLES
# ============================================================

# Real disposable income
df["real_disposable_income"] = (
    df["disposable_income_s14_harmonized"]
    / df["cpi_base_2017_harmonized"]
    * 100
)

# Real consumption
df["real_consumption"] = (
    df["consumption_s14_harmonized"]
    / df["cpi_base_2017_harmonized"]
    * 100
)

# Annual inflation rate
df["inflation"] = (
    df["cpi_base_2017_harmonized"]
    .pct_change()
    * 100
)

# Growth rate of real disposable income
df["real_disposable_income_growth"] = (
    df["real_disposable_income"]
    .pct_change()
    * 100
)

# Growth rate of real consumption
df["real_consumption_growth"] = (
    df["real_consumption"]
    .pct_change()
    * 100
)

# Saving rate
df["saving_rate"] = (
    (
        df["disposable_income_s14_harmonized"]
        - df["consumption_s14_harmonized"]
    )
    / df["disposable_income_s14_harmonized"]
    * 100
)

print("Analytical variables created successfully.")
display(df.head())
```

```python
# ============================================================
# ROUND ANALYTICAL VARIABLES
# ============================================================

df["real_disposable_income"] = (
    df["real_disposable_income"].round(2)
)

df["real_consumption"] = (
    df["real_consumption"].round(2)
)

df["inflation"] = (
    df["inflation"].round(2)
)

df["real_disposable_income_growth"] = (
    df["real_disposable_income_growth"].round(2)
)

df["real_consumption_growth"] = (
    df["real_consumption_growth"].round(2)
)

df["saving_rate"] = (
    df["saving_rate"].round(2)
)

display(df)
```

```python
# ============================================================
# FINAL VALIDATION
# ============================================================

print("Dataset shape:", df.shape)

print("\nColumns:")
for column in df.columns:
    print("-", column)

print("\nMissing values:")
print(df.isna().sum())

display(df.describe().T)
```

```python
# ============================================================
# EXPORT ANALYTICAL DATASET
# ============================================================

output_path = (
    "../data/processed/"
    "study_02_household_purchasing_power_analytical_2007_2024.xlsx"
)

df.to_excel(
    output_path,
    index=False
)

print("Analytical dataset exported successfully.")
print("File:", output_path)
```
