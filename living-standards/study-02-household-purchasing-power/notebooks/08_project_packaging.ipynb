```python
# ============================================================
# 08 - PROJECT PACKAGING
# ============================================================

from pathlib import Path
import os
```

```python
# ============================================================
# PROJECT STRUCTURE CHECK
# ============================================================

required_paths = [
    "../data/raw",
    "../data/processed",
    "../notebooks",
    "../outputs",
    "../outputs/figures",
    "../outputs/tables",
    "../reports"
]

for path in required_paths:

    exists = Path(path).exists()

    status = (
        "OK"
        if exists
        else "MISSING"
    )

    print(
        f"{status:<10} {path}"
    )
```

```python
# ============================================================
# NOTEBOOK CHECK
# ============================================================

required_notebooks = [
    "02_feature_engineering.ipynb",
    "03_exploratory_analysis.ipynb",
    "04_economic_analysis.ipynb",
    "05_statistical_analysis.ipynb",
    "06_export_results.ipynb",
    "07_executive_report.ipynb",
    "08_project_packaging.ipynb"
]

notebooks_path = Path("../notebooks")

for notebook in required_notebooks:

    notebook_file = (
        notebooks_path /
        notebook
    )

    status = (
        "OK"
        if notebook_file.exists()
        else "MISSING"
    )

    print(
        f"{status:<10} {notebook}"
    )
```

```python
# ============================================================
# DATA FILE CHECK
# ============================================================

data_path = Path("../data/processed")

required_data_files = [
    "study_02_household_purchasing_power_2007_2024.xlsx",
    "study_02_household_purchasing_power_analytical_2007_2024.xlsx",
    "study_02_household_purchasing_power_enriched_2007_2024.xlsx"
]

for file_name in required_data_files:

    file_path = (
        data_path /
        file_name
    )

    status = (
        "OK"
        if file_path.exists()
        else "MISSING"
    )

    print(
        f"{status:<10} {file_name}"
    )
```

```python
# ============================================================
# FIGURE CHECK
# ============================================================

figures_path = Path(
    "../outputs/figures"
)

figures = list(
    figures_path.glob("*.png")
)

print(
    f"Number of figures: {len(figures)}"
)

for figure in figures:
    print("-", figure.name)
```

```python
# ============================================================
# TABLE CHECK
# ============================================================

tables_path = Path(
    "../outputs/tables"
)

tables = list(
    tables_path.glob("*.xlsx")
)

print(
    f"Number of tables: {len(tables)}"
)

for table in tables:
    print("-", table.name)
```

```python
# ============================================================
# FINAL PROJECT SUMMARY
# ============================================================

print("=" * 60)
print("STUDY 02 - HOUSEHOLD PURCHASING POWER")
print("=" * 60)

print(
    "Study period: 2007-2024"
)

print(
    "Number of years:",
    18
)

print(
    "Source data:",
    "HCP household sector accounts + harmonized CPI"
)

print(
    "Processed data files:",
    len([
        f for f in data_path.glob("*.xlsx")
    ])
)

print(
    "Figures:",
    len(figures)
)

print(
    "Tables:",
    len(tables)
)

print(
    "Notebooks:",
    len(required_notebooks)
)

print("=" * 60)
print("PROJECT PACKAGING CHECK COMPLETE")
print("=" * 60)
```
