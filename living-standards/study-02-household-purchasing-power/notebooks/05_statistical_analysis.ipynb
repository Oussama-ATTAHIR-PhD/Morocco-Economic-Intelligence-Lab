```python
# ============================================================
# 05 - STATISTICAL ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
from scipy.stats import pearsonr

pd.set_option("display.float_format", lambda x: f"{x:,.4f}")

file_path = (
    "../data/processed/"
    "study_02_household_purchasing_power_enriched_2007_2024.xlsx"
)

df = pd.read_excel(file_path)

print("Shape:", df.shape)
```

```python
# ============================================================
# DESCRIPTIVE STATISTICS
# ============================================================

variables = [
    "cpi_base_2017_harmonized",
    "disposable_income_s14_harmonized",
    "consumption_s14_harmonized",
    "real_disposable_income",
    "real_consumption",
    "inflation",
    "real_disposable_income_growth",
    "real_consumption_growth",
    "saving_rate"
]

descriptive_statistics = (
    df[variables]
    .describe()
    .T
    .round(2)
)

display(descriptive_statistics)
```

```python
from pathlib import Path

TABLES_DIR = Path("../outputs/tables")
TABLES_DIR.mkdir(parents=True, exist_ok=True)

descriptive_statistics.to_excel(
    TABLES_DIR / "table_05_descriptive_statistics.xlsx"
)

print("Descriptive statistics exported.")
```

```python
# ============================================================
# CORRELATION MATRIX
# ============================================================

correlation_variables = [
    "inflation",
    "real_disposable_income_growth",
    "real_consumption_growth",
    "saving_rate"
]

correlation_matrix = (
    df[correlation_variables]
    .corr()
    .round(3)
)

display(correlation_matrix)
```

```python
# ============================================================
# PEARSON CORRELATION
# INFLATION VS REAL INCOME GROWTH
# ============================================================

correlation_data = df[
    [
        "inflation",
        "real_disposable_income_growth"
    ]
].dropna()

correlation, p_value = pearsonr(
    correlation_data["inflation"],
    correlation_data["real_disposable_income_growth"]
)

print(
    "Correlation between inflation and real disposable income growth:"
)

print(f"Pearson r = {correlation:.4f}")
print(f"P-value   = {p_value:.4f}")
print(f"Observations = {len(correlation_data)}")
```

```python
# ============================================================
# PEARSON CORRELATION
# INFLATION VS REAL CONSUMPTION GROWTH
# ============================================================

correlation_data_consumption = df[
    [
        "inflation",
        "real_consumption_growth"
    ]
].dropna()

correlation_consumption, p_value_consumption = pearsonr(
    correlation_data_consumption["inflation"],
    correlation_data_consumption["real_consumption_growth"]
)

print(
    "Correlation between inflation and real consumption growth:"
)

print(f"Pearson r = {correlation_consumption:.4f}")
print(f"P-value   = {p_value_consumption:.4f}")
print(
    f"Observations = "
    f"{len(correlation_data_consumption)}"
)
```

```python
# ============================================================
# STATISTICAL RESULTS SUMMARY
# ============================================================

statistical_results = pd.DataFrame({
    "Relationship": [
        "Inflation vs Real Disposable Income Growth",
        "Inflation vs Real Consumption Growth"
    ],
    "Pearson_r": [
        correlation,
        correlation_consumption
    ],
    "P_value": [
        p_value,
        p_value_consumption
    ],
    "Observations": [
        len(correlation_data),
        len(correlation_data_consumption)
    ]
})

statistical_results = statistical_results.round(4)

display(statistical_results)
```

```python
statistical_results.to_excel(
    TABLES_DIR / "table_06_statistical_relationships.xlsx",
    index=False
)

correlation_matrix.to_excel(
    TABLES_DIR / "table_07_correlation_matrix.xlsx"
)

print("Statistical tables exported successfully.")
```
