```python
# ============================================================
# 06 - EXPORT RESULTS
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

OUTPUTS_DIR = Path("../outputs")
TABLES_DIR = OUTPUTS_DIR / "tables"

print("Available output tables:")
print()

for file in TABLES_DIR.glob("*.xlsx"):
    print("-", file.name)
```

```python
# ============================================================
# LOAD MAIN TABLES
# ============================================================

long_term = pd.read_excel(
    TABLES_DIR / "table_01_long_term_evolution.xlsx"
)

periods = pd.read_excel(
    TABLES_DIR / "table_02_period_analysis.xlsx"
)

key_years = pd.read_excel(
    TABLES_DIR / "table_03_key_years.xlsx"
)

descriptive = pd.read_excel(
    TABLES_DIR / "table_05_descriptive_statistics.xlsx"
)

statistics = pd.read_excel(
    TABLES_DIR / "table_06_statistical_relationships.xlsx"
)

display(long_term)
```

```python
# ============================================================
# CONSOLIDATED RESULTS WORKBOOK
# ============================================================

output_file = (
    OUTPUTS_DIR /
    "study_02_household_purchasing_power_results.xlsx"
)

with pd.ExcelWriter(output_file) as writer:

    long_term.to_excel(
        writer,
        sheet_name="Long Term Evolution",
        index=False
    )

    periods.to_excel(
        writer,
        sheet_name="Period Analysis",
        index=False
    )

    key_years.to_excel(
        writer,
        sheet_name="Key Years",
        index=False
    )

    descriptive.to_excel(
        writer,
        sheet_name="Descriptive Statistics"
    )

    statistics.to_excel(
        writer,
        sheet_name="Statistical Analysis",
        index=False
    )

print("Consolidated results exported successfully.")
print(output_file)
```

```python
# ============================================================
# FINAL RESULTS CHECK
# ============================================================

print("Consolidated workbook created:", output_file.exists())
print()

print("Figures:")
for file in (OUTPUTS_DIR / "figures").glob("*.png"):
    print("-", file.name)

print()

print("Tables:")
for file in TABLES_DIR.glob("*.xlsx"):
    print("-", file.name)
```
