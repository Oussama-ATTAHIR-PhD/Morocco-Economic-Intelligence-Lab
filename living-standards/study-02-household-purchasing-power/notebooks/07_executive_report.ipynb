```python
# ============================================================
# 07 - EXECUTIVE REPORT
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

file_path = (
    "../data/processed/"
    "study_02_household_purchasing_power_enriched_2007_2024.xlsx"
)

df = pd.read_excel(file_path)

TABLES_DIR = Path("../outputs/tables")

display(df.head())
```

```python
# ============================================================
# EXECUTIVE KEY FINDINGS
# ============================================================

start = df.iloc[0]
end = df.iloc[-1]

cpi_change = (
    (
        end["cpi_base_2017_harmonized"]
        / start["cpi_base_2017_harmonized"]
        - 1
    )
    * 100
)

nominal_income_change = (
    (
        end["disposable_income_s14_harmonized"]
        / start["disposable_income_s14_harmonized"]
        - 1
    )
    * 100
)

nominal_consumption_change = (
    (
        end["consumption_s14_harmonized"]
        / start["consumption_s14_harmonized"]
        - 1
    )
    * 100
)

real_income_change = (
    (
        end["real_disposable_income"]
        / start["real_disposable_income"]
        - 1
    )
    * 100
)

real_consumption_change = (
    (
        end["real_consumption"]
        / start["real_consumption"]
        - 1
    )
    * 100
)

print("KEY FINDINGS")
print("=" * 50)

print(
    f"CPI change (2007-2024): "
    f"{cpi_change:.2f}%"
)

print(
    f"Nominal disposable income change: "
    f"{nominal_income_change:.2f}%"
)

print(
    f"Nominal consumption change: "
    f"{nominal_consumption_change:.2f}%"
)

print(
    f"Real disposable income change: "
    f"{real_income_change:.2f}%"
)

print(
    f"Real consumption change: "
    f"{real_consumption_change:.2f}%"
)
```

```python
# ============================================================
# KEY YEARS
# ============================================================

highest_inflation = df.loc[
    df["inflation"].idxmax()
]

lowest_real_income_growth = df.loc[
    df["real_disposable_income_growth"].idxmin()
]

highest_real_income_growth = df.loc[
    df["real_disposable_income_growth"].idxmax()
]

highest_saving = df.loc[
    df["saving_rate"].idxmax()
]

lowest_saving = df.loc[
    df["saving_rate"].idxmin()
]

print(
    "Highest inflation:",
    int(highest_inflation["year"]),
    "|",
    round(highest_inflation["inflation"], 2),
    "%"
)

print(
    "Lowest real income growth:",
    int(lowest_real_income_growth["year"]),
    "|",
    round(
        lowest_real_income_growth[
            "real_disposable_income_growth"
        ],
        2
    ),
    "%"
)

print(
    "Highest real income growth:",
    int(highest_real_income_growth["year"]),
    "|",
    round(
        highest_real_income_growth[
            "real_disposable_income_growth"
        ],
        2
    ),
    "%"
)

print(
    "Highest saving rate:",
    int(highest_saving["year"]),
    "|",
    round(highest_saving["saving_rate"], 2),
    "%"
)

print(
    "Lowest saving rate:",
    int(lowest_saving["year"]),
    "|",
    round(lowest_saving["saving_rate"], 2),
    "%"
)
```

```python
# ============================================================
# EXECUTIVE FINDINGS TABLE
# ============================================================

executive_findings = pd.DataFrame({
    "Finding": [
        "CPI increase",
        "Nominal disposable income increase",
        "Nominal consumption increase",
        "Real disposable income increase",
        "Real consumption increase",
        "Highest inflation",
        "Lowest real income growth",
        "Highest real income growth",
        "Highest saving rate",
        "Lowest saving rate"
    ],
    "Value": [
        f"{cpi_change:.2f}%",
        f"{nominal_income_change:.2f}%",
        f"{nominal_consumption_change:.2f}%",
        f"{real_income_change:.2f}%",
        f"{real_consumption_change:.2f}%",
        f"{int(highest_inflation['year'])} ({highest_inflation['inflation']:.2f}%)",
        (
            f"{int(lowest_real_income_growth['year'])} "
            f"({lowest_real_income_growth['real_disposable_income_growth']:.2f}%)"
        ),
        (
            f"{int(highest_real_income_growth['year'])} "
            f"({highest_real_income_growth['real_disposable_income_growth']:.2f}%)"
        ),
        (
            f"{int(highest_saving['year'])} "
            f"({highest_saving['saving_rate']:.2f}%)"
        ),
        (
            f"{int(lowest_saving['year'])} "
            f"({lowest_saving['saving_rate']:.2f}%)"
        )
    ]
})

display(executive_findings)
```

```python
executive_findings.to_excel(
    TABLES_DIR / "table_08_executive_findings.xlsx",
    index=False
)

print("Executive findings exported successfully.")
```
